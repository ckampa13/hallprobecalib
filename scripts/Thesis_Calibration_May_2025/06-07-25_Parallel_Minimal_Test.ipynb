# Main

```python
from joblib import Parallel, delayed
import time
```

```python
def square(x):
    time.sleep(0.5)  # simulate some work
    return x * x
```

```python
inputs = list(range(10))  # 0 through 9
print("Starting parallel job...")
results = Parallel(n_jobs=4)(delayed(square)(x) for x in inputs)
print("Results:", results)
```
